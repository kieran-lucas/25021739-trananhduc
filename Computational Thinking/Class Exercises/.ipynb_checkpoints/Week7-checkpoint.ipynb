```python
#W7A1
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

arr = list(map(int, input().split()))
left, right = 0, len(arr) - 1
target = int(input())
print(binary_search(arr, left, right, target))

    
```

```python
#W7A2
def count_occurrences(arr, x):
    s = 0
    for k in arr:
        if k == x:
            s +=1
    return s

arr = list(map(int, input().split()))
x = int(input())
print(count_occurrences(arr, x))
```

```python
#W7A3 - Bubble Sort
def bubble_sort(arr):
    is_sorted = True
    for i in range(len(arr) - 1):
        for j in range(len(arr) - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                check = False
        if is_sorted:
            break
    return arr

arr = list(map(int, input().split()))
print(*bubble_sort(arr))
            
```

```python
#W7A3 - Selection Sort
def selection_sort(arr):
    for i in range(len(arr) - 1):
        min_index = i
        for j in range(i + 1, len(arr)):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr

arr = list(map(int, input().split()))
print(*selection_sort(arr))
            
```

```python
#W7A4
def find_max_occurrences(arr):
    if not arr:
        return "Mảng rỗng"
    freq = {}
    for x in arr:
        freq[x] = freq.get(x, 0 ) + 1
    max_count = 0
    for count in freq.values():
        if count > max_count:
            max_count = count
    max_value = None
    for x in arr:
        if freq[x] == max_count:
            max_value = x
            break
    return f"{max_value} xuat hien nhieu nhat, som nhat, {max_count} lan"

arr = list(map(int, input().split()))
print(find_max_occurrences(arr))

  
```

```python
#W7A5
def count_pairs(arr, x):
    s = 0
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] + arr[j] == x:
                s += 1
    return s

arr = list(map(int, input().split()))
x = int(input())
print(count_pairs(arr, x))
```

```python
#W7A6
def flatten(arr):
    flatted = []
    for sublist in arr:
        for subitem in sublist:
            flatted.append(subitem)
    return flatted

arr = eval(input())
print(flatten(arr))
```

```python
#W7A7
def LIS(arr):
    dp = [1]*len(arr)
    for i in range(1, len(arr)):
        for j in range(i):
            if arr[j] < arr[i]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)

arr = eval(input())
print(LIS(arr))
```

```python
#W7A8
import heapq
def min_Interval(intervals, queries):
    intervals.sort(key = lambda x : x[0])
    sorted_queries = sorted((q, i) for i, q in enumerate(queries))
    res = [-1]*len(queries)
    heap = []
    i = 0
    for q, idx in sorted_queries:
        while i < len(intervals) and intervals[i][0] <= q:
            l, r = intervals[i]
            heapq.heappush(heap, (r - l + 1, r))
            i += 1
        while heap and heap[0][1] < q:
            heapq.heappop(heap)
        if heap:
            res[idx] = heap[0][0]
        else:
            res[idx] = - 1
    return res

intervals = eval(input())
queries = eval(input())
print(min_Interval(intervals, queries))
    
    
```

```python
#W7A9
def smaller_number(arr):
    s = [0]*len(arr)
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[j] < arr[i]:
                s[i] += 1
    return s

arr = eval(input())
print(smaller_number(arr))

    
```

```python
#W7A10
import bisect
def max_subarray_sum_mod_m(nums, m):
    prefix = 0
    max_sum = 0
    prefix_list = []
    for num in nums:
        prefix = (prefix + num) % m
        max_sum = max(max_sum, prefix)
        idx = bisect.bisect_right(prefix_list, prefix)
        if idx < len(prefix_list):
            candidate = (prefix - prefix_list[idx] + m) % m
            max_sum = max(max_sum, candidate)
        bisect.insort(prefix_list, prefix)
    return max_sum

nums = eval(input())
m = int(input())
print(max_subarray_sum_mod_m(nums, m))
    
```

```python
#W7A11
s = input()
key = input()
clean = ""
for ch in s:
    if ch.isalpha() or ch == " ":
        clean += ch
    else:
        clean += " "
count = 0
words = clean.split()
for w in words:
    if w == key:
        count += 1
print(count)
```

```python
#W7A13
arr = list(map(int, input().split()))
freq = {}
for x in arr:
    freq[x] = freq.get(x, 0) + 1
max_len = 0
for x in freq:
    if x + 1 in freq:
        max_len = max(max_len, freq[x] + freq[x + 1])
print(max_len)
        
```
