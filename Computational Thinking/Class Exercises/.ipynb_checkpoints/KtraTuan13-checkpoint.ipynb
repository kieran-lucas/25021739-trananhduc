```python
Cho một danh sách các số nguyên nums, hãy sắp xếp lại theo thứ tự tăng dần tổng các chữ số,
nếu có tổng các chữ số bằng nhau, sắp xếp theo thứ tự tăng dần
```

```python
def sum_digits(n):
    n = abs(n)
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total
def sort_nums(nums):
    return sorted(nums, key = lambda x : (sum_digits(x), x))

nums = list(map(int, input().split()))
print(sort_nums(nums))
```

```python
Yêu cầu:
Nhập một danh sách các số nguyên (có thể âm).
Sắp xếp các số theo:
- Tổng chữ số 
- Nếu bằng nhau, sắp xếp theo số lượng chữ số (ít chữ số trước).
- Nếu vẫn bằng nhau, sắp xếp theo giá trị thực.

Ví dụ:
Input: 21 3 -12 111
Output: 3 -12 21 111
```

```python
def sum_digits(n):
    total = 0
    n = abs(n)
    while n > 0:
        total += n % 10
        n //= 10
    return total
def count_digits(n):
    if n == 0:
        return 1
    count = 0
    n = abs(n)
    while n > 0:
        count += 1
        n //= 10
    return count
def sort_nums(nums):
    return sorted(nums, key = lambda x : (sum_digits(x), count_digits(x), x))

nums = list(map(int, input().split()))
print(sort_nums(nums))
    
```

```python
Viết hàm đệ quy đếm số chữ số
```

```python
def count_digits(n):
    n = abs(n)
    if n < 10:
        return 1
    return 1 + count_digits(n // 10)

n = int(input())
print(count_digits(n))
```

```python
Viết hàm tạo dict ánh xạ key -> tổng giá trị của key cùng tên
```

```python
def sum_pairs(n):
    if n == 0:
        return ""
    result = {}
    for _ in range(n):
        line = input().split()
        key, value = line[0], int(line[1])
        if key in result:
            result[key] += value
        else:
            result[key] = value
    return result

n = int(input())
print(sum_pairs(n))

        
```
