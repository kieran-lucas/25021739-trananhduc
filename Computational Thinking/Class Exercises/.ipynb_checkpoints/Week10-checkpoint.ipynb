```python
#W9A1
arr1 = list(map(int, input().split()))
arr2 = list(map(int, input().split()))
arr = []
i, j = 0, 0
while i < len(arr1) and j < len(arr2):
    if arr1[i] <= arr2[j]:
        arr.append(arr1[i])
        i += 1
    else:
        arr.append(arr2[j])
        j += 1
if i < len(arr1):
    arr.extend(arr1[i: len(arr1)])
if j < len(arr2):
    arr.extend(arr2[j : len(arr2)])
print(*arr)
    
    
```

```python
#W9A2
import heapq
def kth_matrix(matrix, k):
    if not matrix or not matrix[0] or k <= 0:
        return None
    n, m = len(matrix), len(matrix[0])
    if n*m < k:
        return None
    heap = []
    for i in range(n - 1, - 1, - 1):
        for j in range(m - 1, - 1, - 1):
            if len(heap) < k:
                heapq.heappush(heap, matrix[i][j])
            elif matrix[i][j] > heap[0]:
                heapq.heappushpop(heap, matrix[i][j])
            else:
                break
    return heap[0]

matrix = []
n = int(input())
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
k = int(input())
print(kth_matrix(matrix, k))
```

```python
#W9A3
num = input()
digits = list(num)
min_num = int(num)
for i in range(len(digits)):
    for j in range(i + 1, len(digits)):
        digits[i], digits[j] = digits[j], digits[i]
        candidate = int("".join(digits))
        if candidate < min_num:
            min_num = candidate
        digits[i], digits[j] = digits[j], digits[i]
if min_num == int(num):
    digits[-2], digits[-1] = digits[-1], digits[-2]
    min_num = int("".join(digits))
print(min_num)
```

```python
#W9A6
def permutation(arr):
    n = len(arr)
    i = n - 2
    while i >= 0 and arr[i] >= arr[i + 1]:
        i -= 1
    if i < 0:
        return list(range(1, n + 1))
    j = n - 1
    while arr[j] <= arr[i]:
        j -= 1
    arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1:] = list(reversed(arr[i + 1:]))
    return arr
```

```python
#W9A7
def canPlaceFlowers(flowerbed, k):
    for i in range(len(flowerbed)):
        if flowerbed[i] == 0:
            left = i == 0 or flowerbed[i - 1] == 0
            right = i == len(flowerbed) - 1 or flowerbed[i + 1] == 0
            if left and right:
                flowerbed[i] = 1
                k -= 1
                if k == 0:
                    return True
    return False

flowerbed = list(map(int, input().split()))
k = int(input())
print(canPlaceFlowers(flowerbed, k))
        
```

```python
#W9A8
def is_path_crossing(moves):
    visited = {(0, 0)}
    x, y = 0, 0
    for ch in moves:
        if ch == "R":
            x += 1
        elif ch == "L":
            x -= 1
        elif ch == "U":
            y += 1
        elif ch == "D":
            y -= 1
        if (x, y) in visited:
            return True
        visited.add((x, y))
        
    return False

moves = input()
print(is_path_crossing(moves))
        
            
```

```python
#W9A9
def median(nums):
    a, b, c, d, e = nums
    if a > b: 
        a, b = b, a
    if c > d:
        c, d = d, a
    if a < c:
        a = e
        if a > b:
            a, b = b, a
    else:
        c = e
        if c > d:
            c, d = d, c
    if a < c:
        return min(b, c)
    else:
        return min(a, d)

nums = list(map(int, input().split()))
print(median(nums))

            
    
```

```python
#W9A10
def find_radius(houses, heaters):
    houses.sort()
    heaters.sort()
    max_dist = 0
    n = len(heaters)
    j = 0
    for h in houses:
        while j + 1 < n and abs(heaters[j + 1] - h) <= abs(heaters[j] - h):
            j += 1
        current_dist = abs(heaters[j] - h)
        if current_dist > max_dist:
            max_dist = current_dist
    return max_dist

houses = list(map(int, input().split()))
heaters = list(map(int, input().split()))
print(find_radius(houses, heaters))
```
