```python
#W7A16
arr = list(map(int, input().split()))
freq = {}
for x in arr:
    freq[x] = freq.get(x, 0) + 1
s = 0
for x in freq:
    if x == 0:
        s += freq[x]
    elif x < freq[x]:
        s += freq[x] - x
    elif x > freq[x]:
        s += freq[x]
if s == 0:
    print("YES")
else:
    print("NO", s)
```

```python
#W7A17
arr = list(map(int, input().split()))
candidate = None
count = 0 
for x in arr:
    if count == 0:
        candidate = x
        count = 1
    elif candidate == x:
        count += 1
    else:
        count -= 1
if candidate is not None and arr.count(candidate) > len(arr) // 2:
    print(candidate)
```

```python
#W7A18
arr = list(map(int, input().split()))
even = []
odd = []
for x in arr:
    if x % 2 == 0:
        even.append(x)
    else:
        odd.append(x)
even.sort()
odd.sort(reverse = True)
print(*even, *odd)
```

```python
#W7A19
arr_1 = list(map(int, input().split()))
arr_2 = list(map(int, input().split()))
is_increasing = len(arr_1) < 2 or arr_1[0] <= arr_1[-1]
i = j = 0
res = []
while i < len(arr_1) and j < len(arr_2):
    if is_increasing:
        if arr_1[i] <= arr_2[j]:
            res.append(arr_1[i])
            i += 1
        else:
            res.append(arr_2[j])
            j += 1
    else:
        if arr_1[i] >= arr_2[j]:
            res.append(arr_1[i])
            i += 1
        else:
            res.append(arr_2[j])
            j += 1
if i < len(arr_1):
    res.extend(arr_1[i : len(arr_1)])
if j < len(arr_2):
    res.extend(arr_2[j : len(arr_2)])
print(*res)
    
```

```python
#W7A20
arr = list(map(int, input().split()))
k = int(input())
s = 0
for x in arr:
    if 10**(k - 1) <= x <= 10**k - 1:
        s += x
print(s)
```

```python
#W7A22
n = int(input())
freq = {}
for i in range(n):
    names = input().split()
    name = names[-1]
    freq[name] = freq.get(name, 0) + 1
max_count = 0
max_name = ""
for x, v in freq.items():
    if v > max_count:
        max_count = v
        max_name = x
    elif v == max_count:
        if x < max_name:
            max_name = x
print(max_name)

        
    
```

```python
#W7A23
s = input()
s_1 = ""
for ch in s:
    if ch.isalpha() or ch == " ":
        s_1 += ch
my_list = s_1.split()
count = 0
for x in my_list:
    if len(x) >= 2 and x[0].isupper():
        count += 1
print(count)
```

```python
#W7A25
num = input()
n = len(num)
k = int(input())
stack = []
to_remove = n - k
for digit in num:
    while stack and to_remove > 0 and stack[-1] < digit:
        stack.pop()
        to_remove -= 1
    stack.append(digit)
print("".join(stack[:k]))
      
```

```python
#W7A26
s = input()
num = ""
nums = []
for x in s:
    if x.isdigit():
        num += x
    else:
        if num != "":
            nums.append(int(num))
            num = ""
if num != "":
    nums.append(int(num))
print(sum(nums), f"{sum(nums)/len(nums):.2f}")
```

```python
#W7A27
pairs = {"{" : "}", "[" : "]", "(" : ")"}
stack = []
s = input()
for x in s:
    if x in pairs:
        stack.append(x)
    else:
        if not stack or x != pairs[stack[-1]]:
            print("NO")
            break
        stack.pop()
else:
    if stack:
        print("NO")
    else:
        print("YES")
```

```python
# W7A28
def is_prime(n):
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
def longest_composite(n):
    if n < 4:
        return "Invalid"
    max_len = 0
    max_start = 0
    current_len = 0
    current_start = 0
    for i in range(4, n):
        if not is_prime(i):
            if current_len == 0:
                current_start = i
                current_len = 1
            else:
                current_len += 1
        else:
            if current_len > max_len:
                max_len = current_len
                max_start = current_start
            current_len = 0
    if current_len > max_len:
                max_len = current_len
                max_start = current_start
    return list(range(max_start, max_start + max_len))

n = int(input())
print(*longest_composite(n))
```

```python
#W7A29
arr = list(map(int, input().split()))
max_len = 0
max_start = 0
current_len = 1
current_start = 0
for i in range(1, len(arr)):
    if arr[i] > arr[i - 1]:
        current_len += 1
    else:
        if current_len > max_len:
            max_len = current_len
            max_start = current_start
        current_len = 1
        current_start = i
if current_len > max_len:
    max_len = current_len
    max_start = current_start
print(*arr[max_start : max_start + max_len])
        
```

```python
#W7A30
arr = list(map(int, input().split()))
n = len(arr)
dp = [1]*n
prev = [-1]*n
for i in range(n):
    for j in range(i):
        if arr[j] < arr[i]:
            if dp[j] + 1 > dp[i]:
                dp[i] = dp[j] + 1
                prev[i] = j
            elif dp[j] + 1 == dp[i]:
                if prev[i] == -1 or j < prev[i]:
                    prev[i] = j
                
max_len = max(dp)
idx = dp.index(max_len)
lis = []
while idx != -1:
    lis.append(arr[idx])
    idx = prev[idx]
lis.reverse()
print(*lis)
```

```python
#W7A31
num = input()
digits = list(num)
min_num = int(num)
for i in range(len(num)):
    for j in range(i + 1, len(num)):
        digits[i], digits[j] = digits[j], digits[i]
        candidate = int("".join(digits))
        if candidate < min_num:
            min_num = candidate
        digits[i], digits[j] = digits[j], digits[i]
if min_num == int(num):
    digits[-2], digits[-1] = digits[-1], digits[-2]
    min_num = int("".join(digits))
print(min_num)
```
